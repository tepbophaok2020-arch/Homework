## Exercise 2: Casting values

3. Convert `"42"` to int and store it in `z`.

## Exercise 4: Slicing, modifying, concatenation

Use:

```python
word = "Programming"
first = "Good"
second = "Morning"
```

Tasks:
1. Slice and print `gram` from `word`.
2. Print every second character of `word`.
3. Print `word` in uppercase and lowercase.
4. Replace `"Pro"` with `"Re"` and print result.
5. Concatenate `first` and `second` with a space.

````python
# Write your solution here
word = "Programming"
first = "Good"
second = "Morning"
 
print(word[3:7])
print(word[::2])  # added: every second character (step 2)
print(word.upper())
print(word.lower())
print(word.replace("Pro", "Re"))
print(first + " " + second)
````

## Exercise 5: Formatting and escape characters

1. Create variables `item = "notebook"` and `price = 12.5`.
2. Use an f-string to print: `The notebook costs 12.50 dollars`.
3. Use `.format()` to print the same message.
4. Print a sentence with escaped double quotes around one word.
5. Print text that includes a new line (`\n`) and a tab (`\t`).

````python
# Write your solution here
item = "notebook"
price = 12.5
print(f"The {item} costs {price:.2f} dollars")
print("The {} costs {:.2f} dollars".format(item, price))  # added: .format() (step 3)
print(f"The {item} costs \"{price:.2f}\" dollars")
print(f"The {item} costs\n \"{price:.2f}\" dollars")
print("Name:\tnotebook")  # added: tab example (step 5)
````
